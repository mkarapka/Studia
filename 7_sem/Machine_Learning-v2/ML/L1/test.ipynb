```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.arange(0., 5., 0.2)
plt.plot(t, t**4, 'g-')
plt.show()
```

```python
data = {'a': np.arange(50),
        'c': np.random.randint(0, 50, 50),
        'd': np.random.randn(50)}
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100

plt.scatter('a', 'b', c='c', s='d', data=data)
plt.xlabel('entry a')
plt.ylabel('entry b')
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Wczytanie danych
df = pd.read_csv("data1.csv")

# Tworzymy wykres 3D
fig = px.scatter_3d(
    df,
    x='x1',
    y='x2',
    z='y',
    color='y',           # kolor punktów według wartości targetu
    size_max=10,         # maksymalny rozmiar punktu
    opacity=0.8
)

# Tytuł wykresu
fig.update_layout(title='Interaktywny wykres 3D: x1, x2 vs y')

# Wyświetlenie wykresu
fig.show()
```
